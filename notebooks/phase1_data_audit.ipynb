### **Step 2: Collect La Liga Schedule**

```python
import soccerdata as sd

fbref = sd.FBref(
    leagues="ESP-La Liga",
    seasons=["2025-2026"]
)

schedule = fbref.read_schedule().reset_index()

print(schedule.columns.tolist())
print(schedule.head())
```

### **Step 3: Inspect Key Teams**

```python
teams = ["Real Madrid", "Barcelona", "Atlético Madrid"]

team_matches = schedule[
    schedule["home_team"].isin(teams) |
    schedule["away_team"].isin(teams)
]

print(team_matches[[
    "date",
    "home_team",
    "score",
    "away_team",
    "week"
]].head(20))
```
